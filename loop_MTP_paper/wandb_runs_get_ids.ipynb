```python
import wandb
from tqdm import tqdm

ENTITY = "behzadshomali"
PROJECT = "loop_MTP_paper"
# TAG = "alignment_sweep"

api = wandb.Api()

runs = api.runs(
    f"{ENTITY}/{PROJECT}",
    # filters={"tags": {"$in": [TAG]}},
)

# print(f"Found {len(runs)} runs with tag '{TAG}' in project '{PROJECT}'.")
selected_ids = []
for run in (runs):
    run_name = run.name
    step = run.summary.get('_step', 0)
    if run_name.startswith("TunedAlign"):
        name = run.name.split("]]_")[-1]
        # if run step is more than 27500, print its name
        if step > 27500:
            selected_ids.append((run.id, run.name))
            print(f'"{name}:{run.id}"')
```
