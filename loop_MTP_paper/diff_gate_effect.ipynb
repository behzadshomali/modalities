```python
# ── ACL/EMNLP-friendly style ────────────────────────────────────────────────
plt.rcParams.update({
    "font.family":           "serif",
    "font.serif":            ["Times New Roman", "DejaVu Serif"],
    "mathtext.fontset":      "dejavuserif",
    "font.size":             9,
    "axes.titlesize":        10,
    "axes.labelsize":        9,
    "legend.fontsize":       7,
    "legend.title_fontsize": 7,
    "xtick.labelsize":       8,
    "ytick.labelsize":       8,
    "axes.linewidth":        0.8,
    "xtick.major.width":     0.6,
    "ytick.major.width":     0.6,
    "lines.linewidth":       1.5,
    "figure.dpi":            200,
    "savefig.dpi":           300,
    "savefig.bbox":          "tight",
})

DOMAINS    = ["math", "qa", "code"]
TITLES     = {"math": "Math", "qa": "QA", "code": "Code"}

series = [
    ("LoopMTP (learned gate)",     loop_mtp),
    ("Uniform",                    uniform),
    ("Only last",                  only_last),
    ("Only last (learnable)",      only_last_learnable),
]

PALETTE    = [plt.cm.viridis(v) for v in np.linspace(0.10, 0.80, len(series))]
LINESTYLES = ["-", "--", "-.", ":"]
MARKERS    = ["o", "s", "^", "D"]

fig, axes = plt.subplots(1, 3, figsize=(7.0, 2.2), sharex=True)

for ax, domain in zip(axes, DOMAINS):
    for j, (name, data) in enumerate(series):
        ys = np.array([np.nan if v is None else v for v in data[domain]], dtype=float)
        xs = np.array(loops, dtype=float)
        mask = ~np.isnan(ys)
        if not mask.any():
            continue
        ax.plot(
            xs[mask], ys[mask],
            label=name,
            color=PALETTE[j],
            linestyle=LINESTYLES[j],
            marker=MARKERS[j],
            markersize=4,
            linewidth=1.5,
        )

    ax.set_title(TITLES[domain])
    ax.set_xlabel("Number of Loops")
    ax.set_xticks(loops)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.yaxis.grid(True, linestyle=":", linewidth=0.5, alpha=0.6, color="#aaaaaa")
    ax.set_axisbelow(True)
    ax.xaxis.set_major_locator(ticker.FixedLocator(loops))

axes[0].set_ylabel("BPB ($\\downarrow$)")

handles, labels = [], []
for ax in axes:
    for h, l in zip(*ax.get_legend_handles_labels()):
        if l not in labels:
            handles.append(h)
            labels.append(l)

fig.legend(
    handles, labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.18),
    ncol=len(handles),
    frameon=True,
    framealpha=0.8,
    edgecolor="#cccccc",
)

fig.tight_layout(pad=0.1)
Path("./plots").mkdir(exist_ok=True)
fig.savefig("./plots/diff_gate_effect_bpb.pdf", bbox_inches="tight")
plt.show()
```
