```python
from transformers import AutoTokenizer, AutoModelForCausalLM, TextStreamer
```

```python
model_name = "/raid/s3/opengptx/behzad_shomali/instruction_tuning/PonderingLM_OpenMathInstruct-2/2025_10_13-12_40_47/checkpoint-3000/"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(pretrained_model_name_or_path=model_name, dtype="auto", device_map="auto", trust_remote_code=True)
streamer = TextStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)
```

```python
input_text = "Amy has 2 books. Sara gives him 3 more books. Now Amy has "
inputs = tokenizer(input_text, return_tensors="pt").to(model.device)
```

```python
model.pondering_steps = 20
```

```python

outputs = model.generate(
    **inputs, 
    max_new_tokens=50, 
    do_sample=True, 
    top_p=0.9, 
    temperature=0.7,
    streamer=streamer
)
# print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
