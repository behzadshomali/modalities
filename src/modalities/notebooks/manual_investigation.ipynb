```python
from transformers import AutoModelForCausalLM, AutoTokenizer, TextStreamer

model_name = "Bytedance/Ouro-1.4B-Thinking"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    torch_dtype="auto",
    trust_remote_code=True
)
```

```python
model.config.total_ut_steps = 4
model.config.early_exit_threshold = 0.0001
```

```python
model.config.num_hidden_layers
```

```python
streamer = TextStreamer(tokenizer, skip_special_tokens=True)
messages = [
    {"role": "user", "content": "Solve: If 2x + 3 = 11, what is x?"}
]
inputs = tokenizer.apply_chat_template(
    messages,
    add_special_tokens=True,
    return_tensors="pt"
).to(model.device)

# pass early_exit_threshold = 0.0001 for text generation with early exit
_ = model.generate(
    inputs,
    max_new_tokens=512,
    temperature=1.0,
    top_p=0.7,
    streamer=streamer,
    early_exit_threshold=0.0001
)
```

```python
# Generate with reasoning
messages = [
    {"role": "user", "content": "Solve: If 2x + 3 = 11, what is x?"}
]
inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt"
).to(model.device)
```

```python
outputs = model.generate(inputs, max_new_tokens=512, temperature=1.0, top_p=0.7)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
