```python
# downlaod and load the jsonl file from: https://huggingface.co/datasets/codeparrot/apps/resolve/main/train.jsonl

url = "https://huggingface.co/datasets/codeparrot/apps/resolve/main/train.jsonl"
import requests
response = requests.get(url)
with open("train.jsonl", "wb") as f:
    f.write(response.content)

```

```python
import json
data_list = []
with open("train.jsonl", "r") as f: 
    for i, line in enumerate(f):
        if i < 5000:
            data = json.loads(line)
            data_list.append(data)
```
